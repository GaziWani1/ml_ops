```python
df = pd.read_csv("../data/data.csv")
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace("*", "", regex=False)
)
```

```python
df.head()
```

```python
# Missing values
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
# Duplicate records
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace("*", "", regex=False)
)

print(df.columns.tolist())
```

```python
print("Before:", df.shape)

df = df.drop_duplicates()

print("After:", df.shape)
```

```python
numeric_columns = [
    "year", "month", "day", "hour",
    "PM2.5", "PM10", "SO2", "NO2",
    "CO", "O3", "TEMP", "PRES",
    "DEWP", "RAIN", "WSPM"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Clean categorical columns
for col in ["wd", "station"]:
    df[col] = df[col].astype("string").str.strip()

print(df.dtypes)
```

```python
print("Missing PM2.5:", df["PM2.5"].isnull().sum())

df = df.dropna(subset=["PM2.5"])

print("Shape after removing missing targets:", df.shape)
```

```python
df["datetime"] = pd.to_datetime(
    df[["year", "month", "day", "hour"]],
    errors="coerce"
)

# Remove invalid dates
df = df.dropna(subset=["datetime"])

# Sort records
df = df.sort_values(
    ["station", "datetime"]
).reset_index(drop=True)

df[["datetime", "station", "PM2.5"]].head()
```

```python
df = df.drop(columns=["No"], errors="ignore")
```

```python
print("Final shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nTarget distribution:")
print(df["PM2.5"].describe())
```

```python
OUTPUT_PATH = PROCESSED_DIR / "cleaned_data.csv"

df.to_csv(OUTPUT_PATH, index=False)

print("Preprocessing completed!")
print("Saved to:", OUTPUT_PATH)
```
